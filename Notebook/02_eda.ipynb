```python
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import Markdown, display

warnings.filterwarnings("ignore")
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)
pd.set_option("display.max_columns", None)
```

```python
DATA_PATH = r"E:\Credit Risk Assement System\dataset\raw\archive\application_train.csv"
df = pd.read_csv(DATA_PATH)

print("Shape:", df.shape)
display(df.head())
df.info()
```

```python
num_cols = df.select_dtypes(include=np.number).columns.tolist()
cat_cols = df.select_dtypes(include="object").columns.tolist()
missing_cols = df.columns[df.isnull().any()].tolist()

display(Markdown(f"""
- Rows: **{df.shape[0]:,}**, Columns: **{df.shape[1]}**
- Numerical features: **{len(num_cols)}**
- Categorical features: **{len(cat_cols)}**
- Columns with missing values: **{len(missing_cols)}**
"""))
```

```python
target_counts = df["TARGET"].value_counts()
target_pct = df["TARGET"].value_counts(normalize=True).mul(100).round(2)
display(pd.DataFrame({"count": target_counts, "percent": target_pct}))

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.countplot(x="TARGET", data=df, ax=axes[0])
axes[0].set_title("TARGET Count")
axes[0].set_xlabel("TARGET (0=Repaid, 1=Difficulty)")
axes[0].set_ylabel("Count")

axes[1].pie(target_counts, labels=target_counts.index, autopct="%1.1f%%")
axes[1].set_title("TARGET Proportion")
plt.tight_layout()
plt.show()
```

```python
missing_pct = (df.isnull().sum() / len(df) * 100).round(2)
missing_summary = missing_pct[missing_pct > 0].sort_values(ascending=False)
top20_missing = missing_summary.head(20)

display(top20_missing.to_frame("missing_percent"))

fig, ax = plt.subplots(figsize=(10, 8))
top20_missing.sort_values().plot(kind="barh", ax=ax)
ax.set_title("Top 20 Columns by Missing Percentage")
ax.set_xlabel("Missing %")
ax.set_ylabel("Column")
plt.tight_layout()
plt.show()
```

```python
key_numeric = ["AMT_INCOME_TOTAL", "AMT_CREDIT", "AMT_ANNUITY",
                "AMT_GOODS_PRICE", "DAYS_BIRTH", "DAYS_EMPLOYED"]

for col in key_numeric:
    data = df[col].dropna()
    stats = data.agg(["mean", "median", "std", "min", "max"])
    skew = data.skew()

    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    sns.histplot(data, kde=True, ax=axes[0])
    axes[0].set_title(f"{col} — Distribution")
    axes[0].set_xlabel(col)

    sns.boxplot(x=data, ax=axes[1])
    axes[1].set_title(f"{col} — Boxplot")
    plt.tight_layout()
    plt.show()

    display(stats.to_frame(col).T.assign(skewness=skew))
    tag = "highly skewed" if abs(skew) > 1 else "moderately skewed" if abs(skew) > 0.5 else "roughly symmetric"
    print(f"Observation: {col} is {tag} (skew={skew:.2f}).\n")
```

```python
key_categorical = ["CODE_GENDER", "NAME_EDUCATION_TYPE", "NAME_INCOME_TYPE",
                    "NAME_FAMILY_STATUS", "NAME_HOUSING_TYPE", "OCCUPATION_TYPE"]

for col in key_categorical:
    pct = df[col].value_counts(normalize=True).mul(100).round(2)

    fig, ax = plt.subplots(figsize=(10, 4))
    sns.countplot(y=col, data=df, order=pct.index, ax=ax)
    ax.set_title(f"{col} — Counts")
    ax.set_xlabel("Count")
    plt.tight_layout()
    plt.show()

    display(pct.to_frame("percent"))
    print(f"Observation: most common category is '{pct.idxmax()}' ({pct.max()}%).\n")
```

```python
corr_target = df[num_cols].corr()["TARGET"].drop("TARGET").sort_values(key=abs, ascending=False)
top20_corr = corr_target.head(20)
display(top20_corr.to_frame("correlation_with_TARGET"))

fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(df[top20_corr.index.tolist() + ["TARGET"]].corr(), cmap="coolwarm", center=0, ax=ax)
ax.set_title("Correlation Heatmap — Top 20 Features vs TARGET")
plt.tight_layout()
plt.show()
```

```python
num_vs_target = ["AMT_INCOME_TOTAL", "AMT_CREDIT", "AMT_ANNUITY", "DAYS_BIRTH"]
cat_vs_target = ["NAME_EDUCATION_TYPE", "CODE_GENDER",
                  "NAME_INCOME_TYPE", "NAME_FAMILY_STATUS"]

for col in num_vs_target:
    fig, ax = plt.subplots(figsize=(8, 4))
    sns.boxplot(x="TARGET", y=col, data=df, ax=ax, showfliers=False)
    ax.set_title(f"{col} by TARGET")
    plt.tight_layout()
    plt.show()

for col in cat_vs_target:
    rate = df.groupby(col)["TARGET"].mean().mul(100).round(2).sort_values(ascending=False)
    fig, ax = plt.subplots(figsize=(8, 4))
    rate.plot(kind="bar", ax=ax)
    ax.set_title(f"Default Rate (%) by {col}")
    ax.set_ylabel("TARGET=1 rate (%)")
    plt.tight_layout()
    plt.show()
    print(f"Observation: highest default rate in '{col}' is '{rate.idxmax()}' ({rate.max()}%).\n")
```

```python
outlier_cols = ["AMT_INCOME_TOTAL", "AMT_CREDIT", "AMT_ANNUITY", "DAYS_EMPLOYED"]

for col in outlier_cols:
    q1, q3 = df[col].quantile([0.25, 0.75])
    iqr = q3 - q1
    n_outliers = ((df[col] < q1 - 1.5 * iqr) | (df[col] > q3 + 1.5 * iqr)).sum()

    fig, ax = plt.subplots(figsize=(8, 3))
    sns.boxplot(x=df[col], ax=ax)
    ax.set_title(f"{col} — Outlier Check ({n_outliers} outliers by IQR rule)")
    plt.tight_layout()
    plt.show()
```

```python
skewness = df[num_cols].skew().sort_values(key=abs, ascending=False)
top20_skew = skewness.head(20)
display(top20_skew.to_frame("skewness"))

fig, ax = plt.subplots(figsize=(10, 8))
top20_skew.sort_values().plot(kind="barh", ax=ax)
ax.set_title("Top 20 Most Skewed Numerical Features")
ax.set_xlabel("Skewness")
plt.tight_layout()
plt.show()
```

### 12. Key Insights

| Finding | Impact | Future Action |
|---|---|---|
| Dataset is highly imbalanced (~8% TARGET=1) | Model may favor majority class | Use SMOTE or class weights |
| 67 columns have missing values | Naive models will error or bias | Impute per-column strategy (median/mode/flag) |
| Some columns >50% missing | Little signal, risk of noise | Consider dropping or "missing" indicator flag |
| Income/credit amounts are right-skewed | Linear models mis-fit tails | Log-transform before linear modeling |
| `DAYS_EMPLOYMENT` has anomalous max values | Placeholder for unemployed, skews stats | Recode placeholder as separate flag |
| `DAYS_BIRTH`/`DAYS_EMPLOYMENT` are negative | Confusing for readers/models | Convert to positive years for interpretability |
| Outliers present in income/credit fields | Can distort mean-based stats | Consider capping/robust scaling later |
| No duplicate rows | Clean identity per applicant | No dedup step needed |
| Gender shows different default rates | Potential fairness/bias concern | Audit for disparate impact before deployment |
| Education level correlates with default rate | Useful predictive signal | Keep as engineered categorical feature |
| Family status affects default rate | Useful predictive signal | Encode carefully, watch for small categories |
| Income type varies widely in default rate | Strong candidate feature | Group rare categories to reduce noise |
| Several numeric features weakly correlate with TARGET individually | No single dominant predictor | Favor ensemble/tree models over linear |
| High-cardinality categoricals (e.g. `OCCUPATION_TYPE`) have missing values | Missingness itself may be informative | Treat "missing" as its own category |
| Categorical columns are unencoded | Not model-ready yet | One-hot/target/ordinal encode in preprocessing notebook |
