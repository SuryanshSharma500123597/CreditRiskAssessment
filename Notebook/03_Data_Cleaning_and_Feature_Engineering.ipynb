```python
import warnings
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

warnings.filterwarnings("ignore")
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)
pd.set_option("display.max_columns", None)

print("pandas:", pd.__version__, "| numpy:", np.__version__)
```

```python
DATA_PATH = Path(r"E:\Credit Risk Assement System\dataset\raw\archive\application_train.csv")

try:
    df = pd.read_csv(DATA_PATH)
except FileNotFoundError as exc:
    raise FileNotFoundError(f"Dataset not found at {DATA_PATH}") from exc

original_shape = df.shape
print("Shape:", df.shape)
display(df.head())
display(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
def build_cleaning_report(data: pd.DataFrame) -> pd.DataFrame:
    """Summarize a DataFrame's shape, missingness, and dtype split."""
    num_cols = data.select_dtypes(include=np.number).columns
    cat_cols = data.select_dtypes(include="object").columns
    report = {
        "total_rows": data.shape[0],
        "total_columns": data.shape[1],
        "columns_with_missing": (data.isnull().sum() > 0).sum(),
        "duplicate_rows": data.duplicated().sum(),
        "numerical_columns": len(num_cols),
        "categorical_columns": len(cat_cols),
    }
    return pd.DataFrame([report])

report_before = build_cleaning_report(df)
display(report_before)
```

```python
def find_high_missing_columns(data: pd.DataFrame, threshold: float = 65.0) -> list:
    """Return columns whose missing percentage exceeds `threshold`."""
    missing_pct = data.isnull().sum() / len(data) * 100
    return missing_pct[missing_pct > threshold].index.tolist()

removed_columns = find_high_missing_columns(df, threshold=65.0)
print(f"Removing {len(removed_columns)} columns (>65% missing):")
print(removed_columns)

shape_before_removal = df.shape
df = df.drop(columns=removed_columns)
print(f"Shape before: {shape_before_removal} -> after: {df.shape}")
```

```python
ANOMALY_VALUE = 365243

df["EMPLOYMENT_ANOMALY"] = (df["DAYS_EMPLOYED"] == ANOMALY_VALUE).astype(int)
df["DAYS_EMPLOYED"] = df["DAYS_EMPLOYED"].replace(ANOMALY_VALUE, np.nan)

print("Anomalous rows flagged:", df["EMPLOYMENT_ANOMALY"].sum())
```

```python
def impute_missing(data: pd.DataFrame) -> pd.DataFrame:
    """Median-fill numeric columns, mode-fill categorical columns
    ('Unknown' if a categorical column has no mode)."""
    data = data.copy()
    num_cols = data.select_dtypes(include=np.number).columns
    cat_cols = data.select_dtypes(include="object").columns

    for col in num_cols:
        if data[col].isnull().any():
            data[col] = data[col].fillna(data[col].median())

    for col in cat_cols:
        if data[col].isnull().any():
            mode = data[col].mode()
            data[col] = data[col].fillna(mode.iloc[0] if not mode.empty else "Unknown")

    return data

missing_before = df.isnull().sum().sum()
df = impute_missing(df)
missing_after = df.isnull().sum().sum()

print("Missing values before:", missing_before)
print("Missing values after:", missing_after)
```

```python
def safe_div(numerator, denominator):
    """Element-wise division; returns NaN where denominator is 0."""
    return np.where(denominator == 0, np.nan, numerator / denominator)

df["AGE"] = (-df["DAYS_BIRTH"] / 365).round(1)
df["YEARS_EMPLOYED"] = (-df["DAYS_EMPLOYED"] / 365).round(1)

df["CREDIT_INCOME_RATIO"] = safe_div(df["AMT_CREDIT"], df["AMT_INCOME_TOTAL"])
df["ANNUITY_INCOME_RATIO"] = safe_div(df["AMT_ANNUITY"], df["AMT_INCOME_TOTAL"])
df["GOODS_CREDIT_RATIO"] = safe_div(df["AMT_GOODS_PRICE"], df["AMT_CREDIT"])
df["EMPLOYMENT_AGE_RATIO"] = safe_div(df["YEARS_EMPLOYED"], df["AGE"])
df["INCOME_PER_CHILD"] = safe_div(df["AMT_INCOME_TOTAL"], df["CNT_CHILDREN"] + 1)
df["CREDIT_PER_CHILD"] = safe_div(df["AMT_CREDIT"], df["CNT_CHILDREN"] + 1)
df["ANNUITY_CREDIT_RATIO"] = safe_div(df["AMT_ANNUITY"], df["AMT_CREDIT"])
df["INCOME_PER_FAMILY_MEMBER"] = safe_div(df["AMT_INCOME_TOTAL"], df["CNT_FAM_MEMBERS"])

engineered_features = [
    "AGE", "YEARS_EMPLOYED", "CREDIT_INCOME_RATIO", "ANNUITY_INCOME_RATIO",
    "GOODS_CREDIT_RATIO", "EMPLOYMENT_AGE_RATIO", "INCOME_PER_CHILD",
    "CREDIT_PER_CHILD", "ANNUITY_CREDIT_RATIO", "INCOME_PER_FAMILY_MEMBER",
]

display(df[engineered_features].describe())
```

```python
for col in ["AMT_INCOME_TOTAL", "AMT_CREDIT", "AMT_ANNUITY"]:
    df[f"LOG_{col}"] = np.log1p(df[col])

log_cols = [f"LOG_{c}" for c in ["AMT_INCOME_TOTAL", "AMT_CREDIT", "AMT_ANNUITY"]]
display(df[log_cols].describe())
```

```python
def iqr_outlier_count(data: pd.DataFrame, col: str) -> int:
    """Count outliers in `col` using the 1.5*IQR rule."""
    q1, q3 = data[col].quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    return int(((data[col] < lower) | (data[col] > upper)).sum())

outlier_cols = ["AMT_INCOME_TOTAL", "AMT_CREDIT", "AMT_ANNUITY", "AMT_GOODS_PRICE",
                "AGE", "YEARS_EMPLOYED"] + engineered_features[2:]

outlier_report = pd.DataFrame([
    {"column": col, "n_outliers": iqr_outlier_count(df, col)}
    for col in outlier_cols
]).sort_values("n_outliers", ascending=False)
display(outlier_report)

for col in outlier_cols:
    fig, ax = plt.subplots(figsize=(8, 2.5))
    sns.boxplot(x=df[col], ax=ax)
    ax.set_title(f"{col} — Outlier Check")
    plt.tight_layout()
    plt.show()
```

```python
final_report = build_cleaning_report(df)

print("Duplicate rows:", df.duplicated().sum())
print("Remaining missing values:", df.isnull().sum().sum())
print("Engineered features present:", all(f in df.columns for f in engineered_features))
print("Final shape:", df.shape)
display(df.dtypes.value_counts())
```
