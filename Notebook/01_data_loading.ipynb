```python
missing_count = df.isnull().sum()
missing_count
```

```python
missing_percent = (missing_count / len(df) * 100).round(2)
missing_summary = pd.DataFrame({
    "missing_count": missing_count,
    "missing_percent": missing_percent,
}).sort_values("missing_percent", ascending=False)

missing_summary[missing_summary["missing_count"] > 0]
```

```python
duplicate_count = df.duplicated().sum()
duplicate_count
```

```python
target_counts = df["TARGET"].value_counts()
target_percent = df["TARGET"].value_counts(normalize=True).mul(100).round(2)

pd.DataFrame({"count": target_counts, "percent": target_percent})
```

```python
top_missing = missing_summary[missing_summary["missing_count"] > 0].head(5)
top_missing_lines = "\n".join(
    f"- `{col}`: {row.missing_percent}% missing"
    for col, row in top_missing.iterrows()
)

target_rate = target_percent.get(1, 0.0)

summary = f"""
### Observations

- The dataset has **{df.shape[0]:,} rows** and **{df.shape[1]} columns**.
- There are **{duplicate_count} duplicate rows**.
- **{(missing_count > 0).sum()} columns** contain at least one missing value.
- Top columns by missing percentage:
{top_missing_lines}
- The target is imbalanced: only **{target_rate}%** of clients are
  labeled `TARGET = 1` (payment difficulties).
- No cleaning or preprocessing has been applied — this is the raw dataset
  exactly as loaded.
"""

display(Markdown(summary))
```
